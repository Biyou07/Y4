## Exercise 1: Book Word Analysis

In this exercise, we will analyze the story to extract insights about word usage. We will focus on identifying frequently occurring words, specifically filtering by word length and grammatical categories.

### Objectives

1. **Identify Frequent Words**: Our first task is to write a program that analyzes the text to find the 10 most frequently occurring words that are longer than 5 characters (don't forget to remove the stop words).
   - **Hint**: You may find NLTK's `defaultdict` and `FreqDist` useful for counting word occurrences efficiently.

2. **Analyze Grammatical Categories**: Next, we will extend our analysis to focus on the grammatical categories of words. Specifically, we will identify the 10 most frequently occurring nouns, verbs, and adjectives in the text.
   - This will help us understand the text's focus based on the types of words used.

3. **Visualization**: Finally, we will visualize the top 10 most frequent nouns, verbs, and adjectives using a plot.
   - Visualizing this data will provide a clear representation of the word distribution and help in interpreting the results.

```python
import nltk
import random
from nltk.tokenize import word_tokenize
from nltk import FreqDist, pos_tag
from collections import defaultdict, Counter
import matplotlib.pyplot as plt
from nltk.corpus import gutenberg, stopwords

# Load the "Alice in Wonderland" text from NLTK's Gutenberg corpus
alice_text = gutenberg.raw('carroll-alice.txt')
stop_words = set(stopwords.words('english'))
print("Total characters:", len(alice_text))
print("\nFirst 500 characters:\n", alice_text[:500])


def analyze_book_words():
    """
    Analyzes word frequency and grammatical categories in Alice in Wonderland
    """
  
    alice_words = word_tokenize(alice_text.lower())
    print("Total unique words:", len(set(alice_words)))
    alice_words = [word for word in alice_words if word.isalpha()]  # Keep only alphabetic words
    alice_words = [word for word in alice_words if word not in stop_words]
    alice_words = [word for word in alice_words if len(word) > 5]   
    print("\n=== Exercise 1: Book Word Analysis ===")
    print("Total words (after removing stop words):", len(alice_words))
    print("Total unique words (after removing stop words):", len(set(alice_words)))
    word_freq = FreqDist(alice_words)
    top_words = word_freq.most_common(10)
    print("Top 10 most frequent words:", top_words)

    words_by_pos = defaultdict(list)

    for word, tag in pos_tag(alice_words):
        if tag.startswith("NN"):
            words_by_pos["nouns"].append(word)
        elif tag.startswith("VB"):
            words_by_pos["verbs"].append(word)
        elif tag.startswith("JJ"):
            words_by_pos["adjectives"].append(word)
    # Count the most frequent words in each grammatical category
    top_nouns = FreqDist(words_by_pos["nouns"]).most_common(10)
    top_verbs = FreqDist(words_by_pos["verbs"]).most_common(10)
    top_adjectives = FreqDist(words_by_pos["adjectives"]).most_common(10)

    # Display the results
    print("Top nouns:", top_nouns)
    print("Top verbs:", top_verbs)
    print("Top adjectives:", top_adjectives)

    # Plot the results
    categories = {
        "Nouns": top_nouns,
        "Verbs": top_verbs,
        "Adjectives": top_adjectives
    }

    plt.figure(figsize=(15, 12))

    for index, (category, words) in enumerate(categories.items(), start=1):
        labels = [word for word, count in words]
        counts = [count for word, count in words]

        plt.subplot(3, 1, index)
        plt.bar(labels, counts)
        plt.title("Top 10 " + category)
        plt.xlabel("Words")
        plt.ylabel("Frequency")
        plt.xticks(rotation=45)

    plt.tight_layout()
    plt.show()


analyze_book_words()
```

### Exercise 2: Text Statistics

Write a program that tracks how often the following main characters appear together within the same paragraph: Alice, Queen, King, Rabbit, Hatter, Duchess.

Output: Display character pairs and the frequency of their co-occurrence.

```python
import nltk
import random
from itertools import combinations
from nltk.tokenize import word_tokenize
from nltk import FreqDist, pos_tag
from collections import defaultdict, Counter
import matplotlib.pyplot as plt
from nltk.corpus import gutenberg, stopwords

# Load the "Alice in Wonderland" text from NLTK's Gutenberg corpus
alice_text = gutenberg.raw('carroll-alice.txt')
stop_words = set(stopwords.words('english'))

def analyze_character_cooccurrence():
    """Count character combinations that co-occur within the same paragraph."""
    print("\n=== Exercise 2: Character Co-occurrence Analysis ===")

    main_characters = {'Alice', 'Queen', 'King', 'Rabbit', 'Hatter', 'Duchess'}
    paragraphs = alice_text.split('\n\n')

    # Create every possible character combination containing at least two characters.
    character_combinations = []
    for size in range(2, len(main_characters) + 1):
        character_combinations.extend(combinations(sorted(main_characters), size))

    combination_counts = Counter()

    for paragraph in paragraphs:
        paragraph_words = set(word_tokenize(paragraph.lower()))
        present_characters = {
            character for character in main_characters
            if character.lower() in paragraph_words
        }

        # Count a combination when every character in it appears in the paragraph.
        for combination in character_combinations:
            if set(combination).issubset(present_characters):
                combination_counts[combination] += 1

    most_common = combination_counts.most_common(10)

    print("Top co-occurring combinations:")
    for combination, count in most_common:
        print(f"{combination}: {count}")

    labels = [" + ".join(combination) for combination, count in most_common]
    counts = [count for combination, count in most_common]

    plt.figure(figsize=(12, 6))
    plt.barh(labels[::-1], counts[::-1])
    plt.title("Most Frequent Character Co-occurrences")
    plt.xlabel("Number of paragraphs")
    plt.ylabel("Character combination")
    plt.tight_layout()
    plt.show()

analyze_character_cooccurrence()
```

```python
import nltk
import random
from nltk.tokenize import word_tokenize
from nltk import FreqDist, pos_tag
from collections import defaultdict, Counter
import matplotlib.pyplot as plt
from nltk.corpus import gutenberg, stopwords

# Load the "Alice in Wonderland" text from NLTK's Gutenberg corpus
  # For reproducibility
alice_text = gutenberg.raw('carroll-alice.txt')
# NOTE: The original code used 'text' later, replacing with 'alice_text' for consistency
text = alice_text
stop_words = set(stopwords.words('english'))

# Tokenize the text and keep all words, including stop words
def preprocess_text(text):
    # Tokenize the text
    tokens = word_tokenize(text) # Added word_tokenize to complete the function
    return [token for token in tokens if token.isalpha()]  # Keep only alphabetic tokens

tokens = preprocess_text(text.lower())  # Convert to lowercase for consistency

# Function to track the words that come immediately after a given word (Unigram/Bigram model)
def analyze_next_word(tokens):
    next_word_dict = defaultdict(list)
    # Loop through tokens and collect the next words
    # Core: Store the word following token[i] using token[i] as the key
    for i in range(len(tokens) - 1):
        current_word = tokens[i]
        next_word = tokens[i + 1]
        next_word_dict[current_word].append(next_word)
    return next_word_dict

# Function to track the words that come immediately after a pair of words (Trigram model)
def analyze_next_word_trigram(tokens):
    next_word_trigram_dict = defaultdict(list)
    # Loop through tokens and collect the next word using the pair (token[i], token[i+1]) as the key
    for i in range(len(tokens) - 2):
        current_pair = (tokens[i], tokens[i + 1])
        next_word = tokens[i + 2]
        next_word_trigram_dict[current_pair].append(next_word)
    return next_word_trigram_dict

# Function to track the words that come immediately after three words (Four-gram model)
def analyze_next_word_fourgram(tokens):
    next_word_fourgram_dict = defaultdict(list)
    # Loop through tokens and collect the next word using the triplet (token[i], token[i+1], token[i+2]) as the key
    for i in range(len(tokens) - 3):
        current_triplet = (tokens[i], tokens[i + 1], tokens[i + 2])
        next_word = tokens[i + 3]
        next_word_fourgram_dict[current_triplet].append(next_word)
    return next_word_fourgram_dict


# Sentence generation using predicted next words (Bigram, based on 1 previous word)
def generate_sentence_bigram(start_word, next_word_dict, sentence_length=15):
    sentence = [start_word]
    current_word = start_word
    # Generate the rest of the sentence by predicting the next word based on the *last* word
    while len(sentence) < sentence_length:

        most_common_next_words = Counter(next_word_dict[current_word]).most_common(5)
        next_word = random.choice([word for word, _ in most_common_next_words]) if current_word in next_word_dict else None
        if not next_word:
            break
        sentence.append(next_word)
        current_word = next_word
        

    return ' '.join(sentence)

# Sentence generation using Trigram prediction (based on 2 previous words)
def generate_sentence_trigram(start_pair, next_word_trigram_dict, sentence_length=15):
    sentence = list(start_pair) # start_pair is a tuple/list of two words
    current_pair = start_pair

    # Generate the rest of the sentence by predicting the next word based on the *last two* words
    while len(sentence) < sentence_length:
        most_common_next_words = Counter(next_word_trigram_dict[current_pair]).most_common(5)
        next_word = random.choice([word for word, _ in most_common_next_words]) if current_pair in next_word_trigram_dict else None
        if not next_word:
            break
        sentence.append(next_word)
        current_pair = (current_pair[1], next_word)  # Update the current pair to the last two words

    return ' '.join(sentence[:sentence_length])

# Sentence generation using Four-gram prediction (based on 3 previous words)
def generate_sentence_fourgram(start_triplet, next_word_fourgram_dict, sentence_length=15):
    sentence = list(start_triplet) # start_triplet is a tuple/list of three words
    current_triplet = start_triplet

    # Generate the rest of the sentence by predicting the next word based on the *last three* words
    while len(sentence) < sentence_length:
        most_common_next_words = Counter(next_word_fourgram_dict[current_triplet]).most_common(5)
        next_word = random.choice([word for word, _ in most_common_next_words]) if current_triplet in next_word_fourgram_dict else None
        if not next_word:
            break
        sentence.append(next_word)
        current_triplet = (current_triplet[1], current_triplet[2], next_word)  # Update the current triplet to the last three words

    return ' '.join(sentence[:sentence_length])

# Example usage
start_word = 'alice'
start_pair = ('alice', 'was')
start_triplet = ('alice', 'was', 'beginning')


# Step 1: Analyze next word (Bigram/1-word context) - Replaced old analyze_next_word
next_word_dict = analyze_next_word(tokens)

# Step 2: Analyze next word (Trigram/2-word context) - New for the question
next_word_trigram_dict = analyze_next_word_trigram(tokens)

# Step 3: Analyze next word (Four-gram/3-word context) - New for the question
next_word_fourgram_dict = analyze_next_word_fourgram(tokens)


# Step 4: Generate a sentence using Bigram prediction (1-word context)
print("Generated Sentence (Bigram/1-word Context):")
print(generate_sentence_bigram(start_word, next_word_dict))

# Step 5: Generate a sentence using Trigram prediction (2-word context)
print("\nGenerated Sentence (Trigram/2-word Context):")
print(generate_sentence_trigram(start_pair, next_word_trigram_dict))

# Step 6: Generate a sentence using Four-gram prediction (3-word context)
print("\nGenerated Sentence (Four-gram/3-word Context):")
print(generate_sentence_fourgram(start_triplet, next_word_fourgram_dict))
```

## Exercise 4: Chapter Vocabulary Profiler

In this exercise, you will create a program that evaluates the vocabulary complexity of each chapter in the story. This analysis will provide insights into the linguistic richness of the chapters by calculating various metrics.

### Objectives

1. **Average Word Length**:
   - Calculate the average length of words in each chapter. This metric helps determine the complexity of the vocabulary used.

2. **Number of Unique Words**:
   - Count the number of unique words in each chapter. This indicates the diversity of the vocabulary.

3. **Frequency of Complex Words**:
   - Identify and count the frequency of complex words, defined as words with more than 5 or 6 letters. This metric highlights the use of advanced vocabulary.

4. **Merging Metrics**:
   - Merge the three metrics—average word length, number of unique words, and frequency of complex words—into a single vocabulary score using the following formula:

   Vocabulary Score = (0.4 * Average Word Length) + (0.4 * Number of Unique Words) + (0.2 * Number of Complex Words)

   Apply this formula to generate a combined score for each chapter, allowing for easier comparison between them.


   Apply this formula to generate a combined score for each chapter, allowing for easier comparison between them.

```python
import nltk
from nltk.tokenize import word_tokenize
import matplotlib.pyplot as plt

# Load the "Alice in Wonderland" text from NLTK's Gutenberg corpus
from nltk.corpus import gutenberg
alice_text = gutenberg.raw('carroll-alice.txt')

def analyze_chapter_vocabulary():
    """Analyze vocabulary complexity of each chapter and sort them by complexity"""
    print("=== Exercise 4: Chapter Vocabulary Analysis ===")
    
    # Split text into chapters (ignoring the header before CHAPTER I)
    chapters = alice_text.split("CHAPTER ")[1:]
    chapter_metrics = []

    for idx, chapter in enumerate(chapters, start=1):
        # Tokenize and keep only alphabetic words in lowercase
        words = [w.lower() for w in word_tokenize(chapter) if w.isalpha()]
        
        # 1. Average Word Length
        avg_word_length = sum(len(w) for w in words) / len(words) if words else 0
        
        # 2. Number of Unique Words
        num_unique_words = len(set(words))
        
        # 3. Frequency of Complex Words (more than 6 letters)
        num_complex_words = sum(1 for w in words if len(w) > 6)
        
        # 4. Combined Vocabulary Score
        vocab_score = (0.4 * avg_word_length) + (0.4 * num_unique_words) + (0.2 * num_complex_words)
        
        chapter_metrics.append({
            "chapter": f"Chapter {idx}",
            "avg_length": avg_word_length,
            "unique_words": num_unique_words,
            "complex_words": num_complex_words,
            "score": vocab_score
        })

    # Sort chapters by vocabulary complexity score (descending)
    sorted_chapters = sorted(chapter_metrics, key=lambda x: x["score"], reverse=True)

    for item in sorted_chapters:
        print(
            f"{item['chapter']:10s} | "
            f"Avg Length: {item['avg_length']:.2f} | "
            f"Unique Words: {item['unique_words']:4d} | "
            f"Complex Words: {item['complex_words']:3d} | "
            f"Score: {item['score']:.2f}"
        )

    # Plot Vocabulary Score per Chapter
    plt.figure(figsize=(12, 6))
    plt.bar([c["chapter"] for c in sorted_chapters], [c["score"] for c in sorted_chapters])
    plt.title("Vocabulary Complexity Score by Chapter")
    plt.xlabel("Chapter")
    plt.ylabel("Vocabulary Score")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

# Call the function
analyze_chapter_vocabulary()
```

## Exercise 5: Character Action Tracker

In this exercise, you will develop a program to identify and track the actions (verbs) most commonly associated with the main characters in the story. This analysis will help you understand the behaviors and activities of the characters throughout the narrative.

### Main Characters

For this analysis, we will focus on the following main characters: **Alice, Queen, King, Rabbit, Hatter, Duchess**.

### Objectives

1. **Identify Actions**:
   - Analyze the text to extract **verbs** associated with each main character. This involves tracking the occurrences of verbs that appear within a short context around the character's name.
   - **Hint**: Focus on identifying verbs that occur **within the next 4 words** after the character's name appears in a sentence.

2. **Count and Rank**:
   - Count how many times each verb is associated with each character and rank them to find the most frequent actions.

3. **Output Results**:
   - For each main character, print a list of the top verbs associated with them, along with their frequencies.


```python
import nltk
import random
from nltk.tokenize import word_tokenize
from nltk import FreqDist, pos_tag
from collections import defaultdict, Counter
import matplotlib.pyplot as plt
from nltk.corpus import gutenberg, stopwords

# Load the "Alice in Wonderland" text from NLTK's Gutenberg corpus
alice_text = gutenberg.raw('carroll-alice.txt')
stop_words = set(stopwords.words('english'))

def track_character_actions():
    """Track character actions using POS tagging"""
    print("=== Exercise 5: Character Action Tracker ===")

    main_characters = ["Alice", "Queen", "King", "Rabbit", "Hatter", "Duchess"]
    char_lookup = {c.lower(): c for c in main_characters}
    character_verbs = defaultdict(Counter)

    # Tokenize text, keep alphabetic words, and POS-tag them
    words = [w.lower() for w in word_tokenize(alice_text) if w.isalpha()]
    tagged_words = pos_tag(words)

    # 1. Identify verbs within the next 4 words after each character's name
    for i, (word, _) in enumerate(tagged_words):
        if word in char_lookup:
            char_name = char_lookup[word]
            for next_word, tag in tagged_words[i + 1 : i + 5]:
                if tag.startswith("VB") and next_word not in stop_words:
                    character_verbs[char_name][next_word] += 1

    # 2 & 3. Rank and output the top verbs for each character
    for character in main_characters:
        top_actions = character_verbs[character].most_common(5)
        print(f"{character:8s}: {top_actions}")

track_character_actions()
```

```python
import nltk
import matplotlib.pyplot as plt
from nltk.tokenize import sent_tokenize, word_tokenize

# Load the "Alice in Wonderland" text from the NLTK Gutenberg corpus
from nltk.corpus import gutenberg
alice_text = gutenberg.raw('carroll-alice.txt')

# Function to check if a word is a noun or pronoun
def is_noun_or_pronoun(word):
    """Return True when the word is tagged as a noun or pronoun."""
    pos_tagged = nltk.pos_tag([word])
    pos = pos_tagged[0][1]
    return pos.startswith('NN') or pos.startswith('PRP')


# Function to extract and categorize sentences
def categorize_sentences(text):
    """Categorize sentences into Questions, Exclamations, Statements, and Imperatives."""
    sentences = sent_tokenize(text)
    categories = {
        'Questions': [],
        'Exclamations': [],
        'Statements': [],
        'Imperatives': []
    }
    question_words = {'who', 'what', 'why', 'where'}

    for sentence in sentences:
        words = word_tokenize(sentence)
        if not words:
            continue

        first_word = words[0].lower()
        first_tag = nltk.pos_tag([first_word])[0][1]

        # Questions must start with a wh-word and end with '?'.
        if first_word in question_words and sentence.rstrip().endswith('?'):
            categories['Questions'].append(sentence)
        # Imperatives usually start with a base-form verb and end with '.' or '!'.
        elif first_tag == 'VB' and sentence.rstrip().endswith(('.', '!')):
            categories['Imperatives'].append(sentence)
        # Exclamations end with '!'.
        elif sentence.rstrip().endswith('!'):
            categories['Exclamations'].append(sentence)
        # Statements start with a noun/pronoun and end with '.'.
        elif (first_tag.startswith('NN') or first_tag.startswith('PRP')) \
                and sentence.rstrip().endswith('.'):
            categories['Statements'].append(sentence)

    return categories


# Categorize the sentences in the text
categories = categorize_sentences(alice_text)

# Function to plot the frequencies of the categories
def plot_category_frequencies(categories):
    """Plot the number of sentences in each category."""
    labels = list(categories.keys())
    frequencies = [len(categories[label]) for label in labels]

    plt.figure(figsize=(8, 5))
    plt.bar(labels, frequencies)
    plt.title('Sentence Type Frequencies')
    plt.xlabel('Sentence type')
    plt.ylabel('Number of sentences')
    plt.xticks(rotation=20)
    plt.tight_layout()
    plt.show()


# Print a sample of each category
for category, sentences in categories.items():
    print(f'\n{category}: {len(sentences)}')
    for sentence in sentences[:5]:
        print('-', sentence)

# Plot the frequencies
plot_category_frequencies(categories)
```

### Exercise 7: Analysis of Violent Actions and Death Mentions

In this exercise, you will develop a program to identify and analyze sentences in the story that contain mentions of violent actions or death. This analysis aims to explore the themes of violence and mortality within the narrative.

### Tasks

1. **Identify Relevant Sentences**:
   - Write a program that scans the text for sentences containing specific keywords related to violent actions or death. Use the following keywords as indicators:
     - **Death-Related Words**:
       - "kill"
       - "murder"
       - "die"
       - "death"
     - **Violent Action Words**:
       - "attack"
       - "stab"
       - "shoot"
       - "injure"
       - "hurt"
       - "blood"
       - "assault"

2. **Categorize Instances**:
   - After identifying sentences containing the relevant keywords, categorize them into two distinct groups:
     - **Death Indicators**: Sentences containing any of the death-related words.
     - **Violent Actions**: Sentences containing any of the violent action words.

3. **Display Results**:
   - For each category, print the top 10 sentences that correspond to each keyword. This will provide insights into how often these themes appear in the text and in what context.


```python
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import gutenberg

alice_text = gutenberg.raw('carroll-alice.txt')

death_related_words = {"kill", "murder", "die", "death"}
violent_action_words = {"attack", "stab", "shoot", "injure", "hurt", "blood", "assault"}

death_sentences = []
violent_sentences = []

for sentence in sent_tokenize(alice_text):
    words = {word.lower() for word in word_tokenize(sentence) if word.isalpha()}

    if words & death_related_words:
        death_sentences.append(sentence)

    if words & violent_action_words:
        violent_sentences.append(sentence)

print("Death Indicators:")
for sentence in death_sentences[:10]:
    print("-", sentence)

print("\nViolent Actions:")
for sentence in violent_sentences[:10]:
    print("-", sentence)
```
