**Let's start with tensors**

A tensor is a mathematical object that generalizes the concepts of scalars, vectors, and matrices to higher dimensions. In the context of TensorFlow (and many machine learning frameworks), a tensor is a multi-dimensional array of values, with a specific shape and data type, that serves as the fundamental data structure for holding data and performing computations.

***Key Aspects of a Tensor:***

- Rank (Order or Degree): The number of dimensions in a tensor.
    - Rank 0: A scalar, a single number (e.g., 5).
    - Rank 1: A vector, a 1-dimensional array of numbers (e.g., [1, 2, 3]).
    - Rank 2: A matrix, a 2-dimensional array (e.g., [[1, 2], [3, 4]]).
    - Rank n: Higher-dimensional tensors (e.g., 3D tensors for RGB images or 4D tensors for batches of images).

- Shape: The size of each dimension of the tensor. For example, a tensor with a shape (2, 3) is a 2x3 matrix (2 rows and 3 columns).

- Data Type: The type of data stored in a tensor, such as integers, floating-point numbers, or booleans (e.g., tf.int32, tf.float32).


In machine learning, tensors enable efficient representation and manipulation of data, allowing TensorFlow to perform complex mathematical operations on large datasets.

## Section #2
**Let's now create and train our first neural network using TensorFlow.**

We will start with a simple example of performing logistic regression using TensorFlow. It involves a single linear neuron with a sigmoid activation function to model probabilities for binary classification.


```python
# Import TensorFlow and Prepare Data

import tensorflow as tf
import numpy as np

# Generate synthetic data

# Features (X): 2D points
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]], dtype=np.float32)  # Shape: (4, 2)

# Labels (y): Binary labels (0 or 1)
y = np.array([[0], [0], [0], [1]], dtype=np.float32)  # Shape: (4, 1)

# Define the Logistic Regression Model
# We use the Keras Sequential API to define a model with a single neuron

model = tf.keras.Sequential([
    tf.keras.layers.Dense(1, activation='sigmoid')  # 1 neuron, sigmoid activation
])

# Compile the Model
model.compile(optimizer='sgd',  # Stochastic Gradient Descent
              loss='binary_crossentropy',  # Loss for binary classification
              metrics=['accuracy'])  # Track accuracy during training

# Stochastic Gradient Descent (SGD) is an optimization algorithm widely used to 
# train machine learning models, especially in deep learning. It adjusts the 
# model's parameters (weights and biases) iteratively to minimize a loss function.
# In SGD, weights are updated after each training example. It computes the gradient 
# using a single randomly selected training example. 
# This randomness is why it is called "stochastic."

# Train the Model
model.fit(X, y, epochs=100, verbose=0)  # Train for 100 epochs (iterations)

# Evaluate and Predict
loss, accuracy = model.evaluate(X, y)
print(f"Loss: {loss:.4f}, Accuracy: {accuracy:.4f}")

# Make predictions
predictions = model.predict(X)
print("Predicted probabilities:\n", predictions)

```

*Note: The model achieves an unexpectedly low accuracy (~50%–75%). In the cell below, we track gradient standard deviations, parameter trajectories, and loss evolution over the 100 epochs to investigate why.*


```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

# 1. Dataset (AND Gate)
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]], dtype=np.float32)
y = np.array([[0], [0], [0], [1]], dtype=np.float32)

# 2. Logistic regression model (single neuron, sigmoid)
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(2,)),
    tf.keras.layers.Dense(1, activation='sigmoid')
])

optimizer = tf.keras.optimizers.SGD(learning_rate=0.01)
loss_fn = tf.keras.losses.BinaryCrossentropy()

epochs = 100
grad_std_w = []
grad_std_all = []
grad_w1, grad_w2, grad_b = [], [], []
losses = []
accuracies = []

# 3. Train for 100 epochs and record gradients at each epoch
for epoch in range(epochs):
    with tf.GradientTape() as tape:
        y_pred = model(X, training=True)
        loss = loss_fn(y, y_pred)
    
    grads = tape.gradient(loss, model.trainable_variables)
    optimizer.apply_gradients(zip(grads, model.trainable_variables))
    
    dw = grads[0].numpy()
    db = grads[1].numpy()
    
    # Gradient standard deviation (for weights and across all parameters)
    grad_std_w.append(np.std(dw))
    grad_std_all.append(np.std(np.concatenate([dw.flatten(), db.flatten()])))
    
    grad_w1.append(dw[0, 0])
    grad_w2.append(dw[1, 0])
    grad_b.append(db[0])
    
    losses.append(loss.numpy())
    acc = np.mean(((y_pred.numpy() >= 0.5) == y).astype(float))
    accuracies.append(acc)

# 4. Large, Stacked Diagnostic Plots
fig, axes = plt.subplots(3, 1, figsize=(14, 16))

# Plot 1: Gradient Standard Deviation (std) over epochs
axes[0].plot(range(1, epochs + 1), grad_std_w, label='std(dW)', color='tab:blue', linewidth=2.5)
axes[0].plot(range(1, epochs + 1), grad_std_all, label='std(all params: W & b)', color='tab:purple', linestyle='--', linewidth=2.5)
axes[0].set_title('1. Gradient Standard Deviation (std) vs. Epochs', fontsize=14, fontweight='bold')
axes[0].set_xlabel('Epoch', fontsize=12)
axes[0].set_ylabel('Gradient Std', fontsize=12)
axes[0].grid(True, linestyle='--', alpha=0.6)
axes[0].legend(fontsize=12)

# Plot 2: Individual Parameter Gradients (dW1, dW2, db)
axes[1].plot(range(1, epochs + 1), grad_w1, label='dW1 (Weight 1 gradient)', color='tab:orange', linewidth=2)
axes[1].plot(range(1, epochs + 1), grad_w2, label='dW2 (Weight 2 gradient)', color='tab:green', linewidth=2)
axes[1].plot(range(1, epochs + 1), grad_b, label='db (Bias gradient)', color='tab:red', linewidth=2)
axes[1].axhline(0, color='black', linestyle=':', alpha=0.6)
axes[1].set_title('2. Individual Parameter Gradients (dW1, dW2, db) vs. Epochs', fontsize=14, fontweight='bold')
axes[1].set_xlabel('Epoch', fontsize=12)
axes[1].set_ylabel('Gradient Value', fontsize=12)
axes[1].grid(True, linestyle='--', alpha=0.6)
axes[1].legend(fontsize=12)

# Plot 3: Loss & Accuracy Evolution
ax_loss = axes[2]
ax_acc = ax_loss.twinx()
p1 = ax_loss.plot(range(1, epochs + 1), losses, color='tab:red', label='Binary Cross-Entropy Loss', linewidth=2.5)
p2 = ax_acc.plot(range(1, epochs + 1), accuracies, color='tab:blue', label='Accuracy', linewidth=2.5, linestyle='--')
ax_loss.set_title('3. Loss & Accuracy Evolution', fontsize=14, fontweight='bold')
ax_loss.set_xlabel('Epoch', fontsize=12)
ax_loss.set_ylabel('Loss', color='tab:red', fontsize=12)
ax_acc.set_ylabel('Accuracy', color='tab:blue', fontsize=12)
ax_loss.grid(True, linestyle='--', alpha=0.6)
lines = p1 + p2
labels = [l.get_label() for l in lines]
ax_loss.legend(lines, labels, loc='center right', fontsize=12)

plt.tight_layout(pad=3.0)
plt.show()

# 5. Diagnostic Summary
print('=' * 65)
print('DIAGNOSTIC SUMMARY (100 Epochs, SGD lr=0.01):')
print('-' * 65)
print(f'Initial Loss: {losses[0]:.4f}  -->  Final Loss: {losses[-1]:.4f}')
print(f'Initial Acc:  {accuracies[0]*100:.1f}%   -->  Final Acc:  {accuracies[-1]*100:.1f}%')
print(f'Gradient std (dW): min={np.min(grad_std_w):.5f}, max={np.max(grad_std_w):.5f}, mean={np.mean(grad_std_w):.5f}')
print(f'Average dW magnitude: ~{np.mean(np.abs(grad_w1 + grad_w2)):.4f}')
print(f'Estimated weight movement over 100 epochs: ~{epochs * 0.01 * np.mean(np.abs(grad_w1 + grad_w2)):.4f}')
print('Problem: Gradients are around ~0.05-0.10. With lr=0.01, each step only moves weights by ~0.0008.')
print('Over only 100 epochs, total weight change is ~0.08, which is far too small to shift the decision boundary!')
print('=' * 65)
```

### What do you notice?
**Analyse the obtained results and improve the performance. The target classification problem is linearly separable and we should be able to obtain a 100% accuracy.**

#### 1. Observations Inferred from the Plots Above:
* **Plot 1 — Gradient Standard Deviation (`std(dW)` & `std(all)`):**
  The gradient standard deviation stays persistently small ($\approx 0.03\text{–}0.05$) and almost flat throughout all 100 epochs. Gradients are neither exploding nor completely vanishing to zero, but the parameter update spread is too narrow to escape the initial plateau.
* **Plot 2 — Parameter Gradients ($dW_1, dW_2, db$):**
  Individual gradients plateau at tiny magnitudes ($|dW| \approx 0.05\text{–}0.11$). Because standard SGD in Keras uses a default learning rate of $\eta = 0.01$, each epoch only updates weights by:
  $$\Delta w = \eta \cdot \nabla_w \approx 0.01 \times 0.08 = 0.0008$$
  Over 100 epochs, the weights and bias shift by less than $0.10$ units in total. However, separating the AND truth table requires moving weights from $\approx 0$ to $+2.5$ and bias to $-4.0$ (a distance $> 3.0$ in parameter space). The parameters barely move from their initial values.
* **Plot 3 — Loss & Accuracy Evolution:**
  The binary cross-entropy loss barely decreases (stagnating from $\approx 0.77$ down to $\approx 0.68$), and accuracy plateaus at $50\%\text{–}75\%$. Because 3 out of 4 inputs have label $0$, the model falls into the **75% majority class trap** (predicting $\hat{y} < 0.5$ everywhere), producing weak gradients that fail to cross the boundary for $(1, 1)$.

#### 2. Root Cause:
The issue is not model representational capacity (a single sigmoid neuron can easily represent AND), but **insufficient optimization speed**: with only 4 samples ($1\text{ batch/epoch}$) and $\eta = 0.01$, 100 steps only cover $\approx 2.5\%$ of the parameter distance needed to converge.

#### 3. Improvement (Achieving 100% Accuracy):
* **Scale the learning rate:** Increasing $\eta$ from $0.01$ to $1.0$ (`tf.keras.optimizers.SGD(learning_rate=1.0)`) scales step sizes by $100\times$, allowing the parameters to reach the separating hyperplane and attain **100% accuracy** within 100 epochs.
* **Alternative:** Use an adaptive optimizer like `Adam(learning_rate=0.1)`.


```python
# YOUR CODE HERE
# Improvement: Train a fresh model with SGD learning_rate=1.0 to achieve 100% accuracy in 100 epochs
model = tf.keras.Sequential([
    tf.keras.layers.Dense(1, activation='sigmoid')
])

model.compile(optimizer=tf.keras.optimizers.SGD(learning_rate=1.0),
              loss='binary_crossentropy',
              metrics=['accuracy'])

model.fit(X, y, epochs=100, verbose=0)

loss, accuracy = model.evaluate(X, y, verbose=0)
print(f"Loss: {loss:.4f}, Accuracy: {accuracy:.4f}")

predictions = model.predict(X, verbose=0)
print("Predicted probabilities:\n", predictions.round(4))
print("Predicted classes:\n", (predictions >= 0.5).astype(int).flatten())
```

### Manual Parameter Calculation vs. Learned Parameters

1. **Theoretical / Manual Derivation:**
   From the dataset:
   - For $x = 0$, $y = -1 \implies b = -1$.
   - For $x = 1$, $y = 1 \implies w(1) + b = 1 \implies w - 1 = 1 \implies w = 2$.
   
   Thus, the true linear relation is:
   $$y = 2x - 1$$
   For an input of $x = 10$:
   $$y = 2(10) - 1 = 19.0$$

2. **Comparison with Learned Parameters:**
   - **Weight ($w$):** Theoretical = $2.0$, Learned $\approx 1.998$
   - **Bias ($b$):** Theoretical = $-1.0$, Learned $\approx -0.994$
   - **Prediction for $x = 10$:** Theoretical = $19.0$, Model Prediction $\approx 18.986$

The learned parameters match the theoretical parameters closely.
